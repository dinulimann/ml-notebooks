```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import kagglehub
```

```python
path = kagglehub.dataset_download("ashaychoudhary/advertising-campaign-performance-dataset")
csv_file = f"{path}/ad_campaign_performance.csv"
df = pd.read_csv(csv_file)
```

```python
print("Informasi Data:")
print(df.info())
print("\nStatistik Deskriptif:")
print(df.describe())
```

```python
print("\nJumlah Data Kosong per Kolom:")
print(df.isnull().sum())
```

```python
plt.figure(figsize=(8, 6))
sns.histplot(df['CTR'], kde=True)  # Ganti 'CTR' sesuai kolom Click-Through Rate
plt.title('Distribusi Click-Through Rate (CTR)')
plt.xlabel('CTR (%)')
plt.ylabel('Frekuensi')
plt.show()
```

```python
plt.figure(figsize=(12, 10))
numeric_df = df.select_dtypes(include=['number'])
sns.heatmap(numeric_df.corr(), annot=True, cmap='coolwarm', fmt=".2f")
plt.title('Matriks Korelasi Antar Variabel Numerik')
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.scatterplot(x='Budget', y='CTR', data=df)
plt.title('Hubungan Pengeluaran Iklan dengan CTR')
plt.xlabel('Pengeluaran Iklan')
plt.ylabel('CTR (%)')
plt.show()
```

```python
plt.figure(figsize=(8, 6))
sns.barplot(x='Platform', y='CTR', data=df)
plt.title('Performa CTR Berdasarkan Platform')
plt.xlabel('Platform')
plt.ylabel('CTR (%)')
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.boxplot(y='Conversion_Rate', data=df)
plt.title('Distribusi Conversion Rate (CR)')
plt.ylabel('Conversion Rate (%)')
plt.show()
```
